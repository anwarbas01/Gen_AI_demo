### **Scatter plot**

```python
import numpy as np

x=np.array([1,2,3])
y=np.array([1,2,3])

plt.scatter(x,y,c='r',marker="v",s=150)
plt.show();
```

### ***Line plot***

```python
x=np.array([1,3,2,5,4])
y=np.array([1,2,3,4,5])
c=['red','blue','green','grey','pink']

plt.plot(x,y,linestyle="dashed",color='c')
plt.scatter(x,y,c='r',marker="*",s=150)
plt.show();
```

### **Bar chart**

```python
import matplotlib.pyplot as plt
year=["1990","2005","2018","2020","2025"]
language=["c","python","java","c#","c++"]

plt.bar(year,language)
plt.show();
```

### **Pie chart**

```python
language=["c","python","java","c#","c++"]
votes=[45,35,18,30,25]
c=['lightgreen','pink','red','skyblue','violet']

plt.pie(votes,labels=language,colors=c)
plt.show();
```

```python
cars = ["Toyota", "BMW", "Audi", "Honda", "Hyundai"]
sales = [60, 20, 15, 25, 10]
exp=[0.05,0,0,0,0]
c=['lightgreen','pink','red','blue','violet']


plt.pie(sales, labels=cars,colors=c,explode=exp)
plt.show();
```

```python
import matplotlib.pyplot as plt

# Sample data (for learning purposes only)
patients = ["Patient 1", "Patient 2", "Patient 3",
            "Patient 4", "Patient 5"]

sugar = [90, 140, 180, 110, 200]
heart_attack = [0, 0, 1, 0, 1]

# Create figure
plt.figure(figsize=(10, 5))

# Plot blood sugar levels
plt.plot(patients, sugar, marker='o',
         color='blue', label='Blood Sugar Level')

# Mark patients with heart attack
for i in range(len(patients)):
    if heart_attack[i] == 1:
        plt.scatter(patients[i], sugar[i],
                    color='red', s=150,
                    marker='X')

# Add title and labels
plt.title("Blood Sugar Levels and Heart Attack Data")
plt.xlabel("Patients")
plt.ylabel("Blood Sugar Level (mg/dL)")
plt.legend()
plt.grid(True)

# Save chart as an image
plt.savefig("heart_attack_sugar.png", dpi=50)

# Display chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sample data (for learning only)
sugar = [80, 90, 100, 110, 120, 130,
         140, 150, 160, 170, 180, 200]

risk_score = [5, 8, 12, 15, 20, 28,
              35, 42, 50, 58, 65, 75]

# Colors for sample categories
colors = [
    'green', 'green', 'green', 'green',
    'orange', 'orange', 'orange', 'orange',
    'red', 'red', 'red', 'red'
]

# Create scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(sugar, risk_score, color=colors, s=100)
```
