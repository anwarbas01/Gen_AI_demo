```python
import os
import csv
import pandas as pd
import matplotlib.pyplot as plt

ip = []
mydict = {}

# data taken from log file

with open("C:\\Users\\anwar\\Downloads\\website_access.log") as f:
    for each_line in f:
        mylist = each_line.split(" ")
        ip.append(mylist[0])

def frequency(mylist):
    for each in mylist:
        if each not in mydict:
            mydict[each] = 1
        else:
            mydict[each] += 1

    return mydict

result = frequency(ip)

print(result)

# covnert to csv file

def write_to_csv(mydict):
    with open("C:\\Users\\anwar\\Downloads\\new_log.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["IP Address", "Frequency"])
        for each in mydict:
            writer.writerow([each, mydict[each]])
write_to_csv(result)
```

```python
#matplot librar

import pandas as pd
import matplotlib.pyplot as plt

# Reading CSV file
df = pd.read_csv("C:\\Users\\anwar\\Downloads\\new_log.csv")
plt.bar(df["IP Address"], df["Frequency"], edgecolor="black")
plt.title("IP ADDRESS AND NUMBER OF REQUESTS")
plt.xlabel("IP ADDRESS")
plt.ylabel("NUMBER OF REQUESTS")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("website_log_chart.png")
plt.show()
```
