```python
def best_k_means(data,kmax):
    means=[]
    intertias=[]
    for k in range(1,kmax):
        kmeans=KMeans(n_clusters=k)
        kmeans.fit(data)
        means.append(k)
        intertias.append(kmeans.interia_)

        
```
