```python
import matplotlib.pyplot as plt

for label in (tele_data.columns):
    plt.hist(tele_data[tele_data['class']=='g'][label],color="pink",label='gamma',alpha=0.5,density=True)
    plt.hist(tele_data[tele_data['class']=='h'][label],color="blue",label='hadron',alpha=0.5,density=True)
    

plt.title(label)
plt.ylabel("probability")
plt.xlabel(label)
plt.legend()
plt.show();
```
