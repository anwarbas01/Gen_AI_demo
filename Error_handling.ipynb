### Error Handling

```python
try:
    a = 10
    b = 2
    print(a/b)
```

```python
try:
    a = 10
    b = 2
    print(a / b)

except:
    print("Error")

else:
    print("No error")
```

```python
while True:
    try:
        num=int(input("enter a number"))
        print(f"the user enter number is = {num}")
        break
    except exception as e:
        print(e)
```

```python
print(20/2)
print(20/0) #zerodivisionerror
```

```python
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    result = a / b

except ValueError:
    print("Please enter numbers only")

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Result =", result)

finally:
    print("Program completed")
```
