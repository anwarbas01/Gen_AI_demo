```python
#Nested loop

row=int(input("enter a row num"))
coloumn=int(input("enter a coloumn num"))

for x in range(row):
    for y in range(1,coloumn+1):
        print(y,end='')
    print()
```

```python
# increasing pattern

n = int(input("Enter number of rows: "))

for i in range(1, n + 1):
    print("* " * i)
```

```python
# decreasing pattern

n = int(input("Enter number of rows: "))

for i in range(n,0, - 1):
    print("$ " * i)
```

```python
#pattern

row=int(input("enter a row"))
symbol=input("enter a symbol")

for i in range(row):
    for j in range(0,i+1):
        print(symbol,end='')
    print()   
```
