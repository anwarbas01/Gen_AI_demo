### ***1. Write a Python program to check whether a given string is a palindrome.***

```python
s = input("Enter a string: ")
if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

### ***2. Write a Python program to find the greatest among four numbers.***

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))
d = int(input("Enter fourth number: "))

if a >= b and a >= c and a >= d:
    print(a, "is greatest")
elif b >= c and b >= d:
    print(b, "is greatest")
elif c >= d:
    print(c, "is greatest")
else:
    print(d, "is greatest")
```

### ***4. Write a Python program to extract uppercase letters, lowercase letters, digits, and special characters from a given string into four separate strings.***

```python
s = input("Enter a string: ")

upper = ""
lower = ""
digits = ""
special = ""

for i in s:
    if i.isupper():
        upper += i
    elif i.islower():
        lower += i
    elif i.isdigit():
        digits += i
    else:
        special += i

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Digits:", digits)
print("Special characters:", special)
```

### ***5. Write a Python program to check whether a given integer is even or odd.***

```python
n = int(input("Enter a number: "))

if n % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```
