```python

import os

def findallfile(name, path):
    mylist = []
    for root, folder, files in os.walk(path):
        if name in files:
            mylist.append(os.path.join(root, name))

    return mylist


for path in findallfile("demo-py.py", "C:\\Users\\anwar\\OneDrive\\Documents\\python"):
    print(path)

print("Search completed")
```
