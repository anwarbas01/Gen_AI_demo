### Loops

***whileloop***
***forloop***

```python
a=1
x=int(input("enter the num")) #rows
y=int(input("enter the num")) #coloumn

while a<=x:
    b=1
    while b<=y:
        print(b,end='')
        b+=1
    print()
    a+=1
print("Loop Over")
```

```python
# while loop

a = 100
while a>=0:
    print(a,end=' ')
    a -=10

print("\nloopover")
```

```python
#For loop

#for variable in sequence:
    # code to execute

# for <variable_name> in <str\list\tuple\set\dict>:
```

```python
#str

for a in "anwar saju": #fetch by letter
   # print(a)
    print(a,end='')
```

```python
#list

mylist=[20,"anwar",2004,"saju",2006,22,"sha"]  #fetch by item by item

for x in mylist:
    print(x)
    #print(x,end=',')
```

```python
#Range

for a in range(2,20+1): #start with end before
    print(a,end=' ')
```

```python
for x in range(0,20+1,2):  #jump
    print(x,end=' ')
```

```python
#Nested loop

row=int(input("enter a row num"))
coloumn=int(input("enter a coloumn num"))

for x in range(row):
    for y in range(1,coloumn+1):
        print(y,end='')
    print()
```

```python
for x in ["color","pen","pencil"]:
    print("------------------")
    for y in ["audi","benz","skoda","bmw"]:
        print(x,y)
    print("--------------")

print("loop over")
```

```python
#pattern

row=int(input("enter a row"))
symbol=input("enter a symbol")

for i in range(row):
    for j in range(0,i+1):
        print(symbol,end='')
    print()          
```
