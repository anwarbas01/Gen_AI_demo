### ***1.Find the length of a list using len()***

```python
mylsit=[1,2,3,4,5,2,4,12,23,34,56]

print(len(mylsit))
```

### *2.max() and min()*

```python
mylist=[10,20,30,40,50]

mylsit=[1,2,3,4,5,2,4,12,23,34,56]

print("Maximum:",max(mylist))
print("Minimum:",min(mylist))
print("-----------------------")
print("Maximum:",max(mylsit))
print("Minimum:",min(mylsit))
```

### *3.calculate total sum()*

```python
mylist=[10,20,30,40,50]

mylsit=[1,2,3,4,5,2,4,12,23,34,56]

print("total:",sum(mylist))
print("-----------------------")
print("total:",sum(mylsit))
```

### *4.list in ascending and descending order using sort()*

```python
mylist=[10,40,20,30,50,60,80]

mylist.sort()
print("Ascending:", mylist)

mylist.sort(reverse=True)
print("Descending:", mylist)
```

### *5.create a tuple and print it*

### *6.to access first,middle and last element from tuple*

```python
mytuple=(10,20,30,40,50)

print("First:", mytuple[0])
print("Middle:", mytuple[len(mytuple) // 2])
print("Last:", mytuple[-1])
```

### *7. Slice a tuple(first 3 and last 3 element)*

```python
mytuple=(10,20,40,50,60,20,40,50)

print("first 3:",mytuple[:3])
print("Last 3:",mytuple[-3:])
```

### *8.Remove an element from a set using discard()*

```python
myset={10,20,40,50,69,70,40}

myset.discard(40)

print(myset)
```

### *9. Remove a random element from a set using pop()*

```python
myset={10,20,30,40,50}

remove=myset.pop()

print("removed set:",remove)
print("newset:",myset)
```

### *10. Update the value of an existing dictionary key*

```python
mydict={"name":"saju", "age":"21", "city":"coimbatore"}

mydict["city"]="trichy"

print(mydict)
```

### *11.remove the last inserted key using pop()*

```python
mydict={"name":"saju", "age":"21", "city":"coimbatore"}

mydict.popitem()

print(mydict)
```

### *12. Remove all key-value pairs using clear()*

```python
mydict={"name":"saju", "age":"21", "city":"coimbatore"}

mydict.clear()  #empty list

print(mydict)
```

### *13.program to get all key using keys()*

### *14.program to get all key-value pair using items()*

### *15.to check if a key exits in a dictionary using in...*

```python
mydict={"name":"saju", "age":"21", "city":"trichy"}

if "age" in mydict:
    print("key exits")

else:
    print("key does not exits")
```
